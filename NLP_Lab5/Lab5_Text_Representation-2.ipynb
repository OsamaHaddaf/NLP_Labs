# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

sentences = [
    "This is the first document.",
    "This document is the second document.",
    "And this is the third one.",
    "Is this the first document?",
]

vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform(sentences)

similarity = cosine_similarity(vectors)
pd.DataFrame(similarity, index=sentences, columns=sentences).round(3)
````

**Observation:** Sentences 1 and 4 (*"This is the first document."* / *"Is this the first document?"*) have a cosine similarity of **1.0** — they contain exactly the same words, and TF-IDF ignores word order and punctuation. Sentence 3 is the least similar to the others because it shares only the common words "this", "is" and "the".

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

sentences = [
    "data science is one of the most important fields of science",
    "this is one of the best data science courses",
    "data scientists analyze data",
]

tfidf = TfidfVectorizer()
result = tfidf.fit_transform(sentences)

tfidf_df = pd.DataFrame(result.toarray(), columns=tfidf.get_feature_names_out(),
                        index=["Sentence 1", "Sentence 2", "Sentence 3"])
tfidf_df.round(3)
````

````python
# The most important words in each sentence (highest TF-IDF score)
for name, row in tfidf_df.iterrows():
    top_words = row.sort_values(ascending=False).head(3)
    print(name, "->", ", ".join(f"{w} ({s:.3f})" for w, s in top_words.items()))
````

**Observation:** Words that appear in every sentence (like **"data"**) get a low IDF, so they are less important — unless they are repeated inside one sentence ("data" appears twice in sentence 3, so it becomes its top word). Words that are unique to one sentence or repeated in it get the highest scores: **"science"** and **"fields"** for sentence 1, **"best"** and **"courses"** for sentence 2, **"analyze"** and **"scientists"** for sentence 3. Note that stop words such as "of" and "this" also score high because they were not removed; removing stop words first (`TfidfVectorizer(stop_words="english")`) would leave only the meaningful words.

````python
import re
import pandas as pd
import gensim
import nltk
from nltk.tokenize import word_tokenize

nltk.download("punkt_tab", quiet=True)

simpsons = pd.read_csv("dataset/simpsons_script_lines.csv")
simpsons = simpsons.dropna(subset=["spoken_words"])
print("Number of lines:", len(simpsons))
simpsons.head()
````

````python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

simpsons["clean_words"] = simpsons["spoken_words"].apply(clean_text)

tokens = [word_tokenize(line) for line in simpsons["clean_words"]]
print(tokens[:3])
````

````python
Skip_gram_model = gensim.models.Word2Vec(tokens, min_count=1, vector_size=100, window=5, sg=1)
print("Vocabulary size:", len(Skip_gram_model.wv))
````
